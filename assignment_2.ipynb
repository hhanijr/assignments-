```python
import numpy as np

# 1. تعريف الدوال الرياضية (Activation Functions)
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

# 2. البيانات (Input Sequence)
X = [1, 2, 3, 4]

# 3. الحالات الابتدائية (Initial States)
# نستخدم float لضمان دقة الحسابات العشرية
h = 0.0  # Hidden State
c = 0.0  # Cell State

# 4. الأوزان (LSTM Parameters)
# أوزان المدخلات (W) وأوزان الحالة المخفية السابقة (U) والانحياز (b)
Wf, Wi, Wo, Wc = 0.4, 0.7, 0.5, 0.6
Uf, Ui, Uo, Uc = 0.1, 0.1, 0.1, 0.1
bf, bi, bo, bc = 0.0, 0.0, 0.0, 0.0

print("--- محاكاة خلية LSTM خطوة بخطوة --- \n")

# 5. حلقة التكرار عبر الزمن (Time Steps)
for t in range(len(X)):
    x = X[t]
    h_prev = h  # الاحتفاظ بـ h من الخطوة السابقة لاستخدامها في كل البوابات
    c_prev = c  # الاحتفاظ بـ c من الخطوة السابقة

    # أ- بوابة النسيان (Forget Gate): تقرر ما سيتم حذفه من الذاكرة القديمة
    f = sigmoid(Wf * x + Uf * h_prev + bf)

    # ب- بوابة الإدخال (Input Gate): تقرر ما سيتم إضافته للذاكرة
    i = sigmoid(Wi * x + Ui * h_prev + bi)

    # ج- الحالة المرشحة (Candidate Cell State): المعلومات الجديدة المقترحة
    c_bar = tanh(Wc * x + Uc * h_prev + bc)

    # د- تحديث حالة الخلية (Cell State Update): دمج القديم والجديد
    c = (f * c_prev) + (i * c_bar)

    # هـ- بوابة المخرجات (Output Gate): تقرر ما سيخرج كـ Hidden State
    o = sigmoid(Wo * x + Uo * h_prev + bo)

    # و- تحديث الحالة المخفية (Hidden State Update)
    h = o * tanh(c)

    # طباعة النتائج لكل خطوة بتنسيق واضح
    print(f"الخطوة الزمنية {t+1}: المدخل (x) = {x}")
    print(f"  - بوابة النسيان (f): {f:.4f}")
    print(f"  - بوابة الإدخال (i): {i:.4f}")
    print(f"  - ذاكرة الخلية (c): {c:.4f}")
    print(f"  - الحالة المخفية (h): {h:.4f}")
    print("-" * 30)

# 6. طبقة التنبؤ النهائية (Dense/Output Layer)
Wy = 1.8
by = 0.3
y_pred = Wy * h + by

print(f"\nالقيمة المتوقعة النهائية (Predicted Next Value) = {y_pred:.3f}")
```
